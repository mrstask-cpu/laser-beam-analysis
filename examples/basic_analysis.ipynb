Minimal end-to-end example.

```python
import numpy as np
from scipy.ndimage import rotate
from laser_beam_analysis import analyze_beam

# Synthetic tilted beam: elongated Gaussian + background + noise
h, w = 300, 300
y, x = np.mgrid[0:h, 0:w]
xc, yc = w / 2.0, h / 2.0
beam = np.exp(-((x - xc) ** 2) / (2 * 12.0 ** 2) - ((y - yc) ** 2) / (2 * 80.0 ** 2))
beam = rotate(beam, 5.0, reshape=False, order=3, mode="constant", cval=0.0)

rng = np.random.default_rng(0)
image = beam * 100.0 + 20.0 + rng.normal(0, 1.5, size=(h, w))

result = analyze_beam(image)

print("Angle (deg):", result.alignment["angle_deg"])
print("Background: ", result.preprocessing["background"])
print("Symmetry correlation:", result.symmetry["correlation"])
print("BASEX zR (px):", result.propagation["basex"]["zR"])
print("HL zR (px):   ", result.propagation["hansenlaw"]["zR"])
print("Method mean difference (%):", result.validation["method_comparison"]["mean_difference_percent"])
print("Audit axis RMS (px):", result.audit["axis_rms_px"])
```
